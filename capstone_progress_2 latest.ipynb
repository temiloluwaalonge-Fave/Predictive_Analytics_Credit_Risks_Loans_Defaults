## Replacing Missing Numerical Values
### Several numerical variables contained missing values and exhibited outliers. Median imputation was used because the median is robust to extreme values and provides a more representative measure of central tendency for skewed financial data. Using the mean could have produced biased imputations due to the presence of outliers.


````python
import numpy as np

# Select numerical columns
num_cols = df.select_dtypes(include=['int64', 'float64']).columns

# Fill missing values with the median
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
````

## Replacing missing categorical Values
#### Several categorical variables contained missing values. Mode imputation was used because the mode represents the most frequently occurring category and preserves the existing distribution of the data. Using the mode provides a simple and reasonable estimate for missing categorical values without introducing invalid or artificial categories.

````python
# Select categorical columns
cat_cols = df.select_dtypes(include='object').columns

# Replace missing values with the most frequent category
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
````

## Interpretation of Tuned Model Results

After hyperparameter tuning, the optimized Logistic Regression, Decision Tree, and Random Forest models were evaluated using the test dataset. The evaluation focused on how effectively each model identified customers who were likely to default while also minimizing incorrect classifications.

---

# Tuned Logistic Regression Results

The tuned Logistic Regression model achieved an accuracy of **68.5%**. This means that the model correctly classified approximately **68.5% of all customers in the test dataset**. However, because the dataset contains significantly more non-default customers than default customers, accuracy alone does not fully represent the model's ability to identify risky borrowers.

For the default class (class 1), the model achieved:

* **Precision: 0.16**
* **Recall: 0.68**
* **F1-score: 0.26**

The precision value of **0.16** means that among all customers predicted by the model as defaulters, only **16% were actually defaulters**. This indicates that the model generated a high number of false positive predictions, where customers who would have repaid their loans were incorrectly classified as high risk.

The recall value of **0.68** means that the model successfully identified **68% of all actual default customers**. From the 4,949 customers who actually defaulted, the model correctly detected **3,350 customers** and failed to identify **1,599 customers**.

The F1-score of **0.26** shows the balance between identifying default customers and avoiding incorrect default predictions. The score indicates that although the model was able to detect a reasonable number of defaults, its low precision reduced its overall effectiveness.

### Build an Artificial Neural Network (ANN)

````python
# Import libraries
import tensorflow as tf
import numpy as np

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

from sklearn.utils import class_weight
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix


# Calculate class weights to handle imbalance
weights = class_weight.compute_class_weight(
    class_weight="balanced",
    classes=np.unique(y_train),
    y=y_train
)

class_weights = {
    0: weights[0],
    1: weights[1]
}

print("Class weights:", class_weights)


# Build ANN model
ann = Sequential([
    Dense(64, activation="relu", input_shape=(X_train_scaled.shape[1],)),
    Dropout(0.3),
    
    Dense(32, activation="relu"),
    
    Dense(1, activation="sigmoid")
])


# Compile model
ann.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)


# Stop training if validation performance stops improving
early_stop = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)


# Train ANN
history = ann.fit(
    X_train_scaled,
    y_train,
    validation_split=0.2,
    epochs=30,
    batch_size=32,
    callbacks=[early_stop],
    class_weight=class_weights,
    verbose=1
)


# Predict probabilities
y_prob = ann.predict(X_test_scaled)


# Convert probabilities into classes
y_pred_ann = (y_prob >= 0.5).astype(int)


# Evaluate model
print("\nAccuracy:")
print(accuracy_score(y_test, y_pred_ann))


print("\nClassification Report:")
print(classification_report(y_test, y_pred_ann))


print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_ann))
````
